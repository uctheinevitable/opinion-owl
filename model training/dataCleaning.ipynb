```python
import re
import spacy
import pandas as pd
import numpy as np
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def clean_text(text):
    # Convert to string
    text = str(text)

    # Remove URLs
    text = re.sub(r"https?://\S+|www\.\S+", "", text)

    # Remove Reddit usernames
    text = re.sub(r"u/\w+", "", text)

    # Remove subreddit mentions
    text = re.sub(r"r/\w+", "", text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Process text with spaCy
    doc = nlp(text)

    # Lemmatize, remove stopwords, punctuation, spaces
    tokens = [
        token.lemma_.lower()
        for token in doc
        if not token.is_stop
        and not token.is_punct
        and not token.is_space
        and token.is_alpha
    ]

    return " ".join(tokens)
```

```python
data = pd.read_csv('../dataset/data/combined_train.csv')
```

```python
data
```

```python
data['text'] = data['text'].apply(clean_text)
```

```python
data['aspect_term'] = data['aspect_term'].apply(clean_text)
```

```python
data['polarity'] = data['polarity'].apply(clean_text)
```

```python
cleanup_map = {'postive': 'positive', 'po': 'positive'}
data['polarity'] = data['polarity'].replace(cleanup_map)
data = data.dropna(subset=['polarity'])
```
