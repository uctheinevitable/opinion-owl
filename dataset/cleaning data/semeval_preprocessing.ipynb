```python
from datasets import load_dataset
import pandas as pd
import numpy as np
ds = load_dataset("tomaarsen/setfit-absa-semeval-laptops")
```

```python
train_df = ds["train"].to_pandas()
test_df = ds["test"].to_pandas()
```

```python
train_df.rename(columns={'span': 'aspect_term', 'label': 'polarity'}, inplace=True)
test_df.rename(columns={'span': 'aspect_term', 'label': 'polarity'}, inplace=True)
```

```python
train_df = train_df.drop(columns=['ordinal'],)
```

```python
test_df = test_df.drop(columns=['ordinal'])
```

```python
train_df.to_csv('../data/semeval_train.csv', index=False)
```

```python
test_df.to_csv('../data/semeval_test.csv', index=False)
```
